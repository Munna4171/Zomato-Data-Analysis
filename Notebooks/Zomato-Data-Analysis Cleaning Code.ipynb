```python
import pandas as pd

# 1. Load the data
df = pd.read_csv('/content/drive/MyDrive/Zomato-data-.csv')

# 2. Clean the 'rate' column
# This removes the '/5' and converts the string to a float
df['rate'] = df['rate'].astype(str).apply(lambda x: x.split('/')[0].strip())
df['rate'] = pd.to_numeric(df['rate'], errors='coerce')

# 3. Rename complex columns for easier handling
df.rename(columns={
    'approx_cost(for two people)': 'cost_for_two',
    'listed_in(type)': 'restaurant_type'
}, inplace=True)

# 4. Save the cleaned data for your dashboard
df.to_csv('zomato_cleaned.csv', index=False)

print("Data cleaned and saved successfully!")
```
